```python
import numpy as np
import pandas as pd
import plotly.express as px
from textblob import TextBlob

df=pd.read_csv('netflix_titles.csv')
df
```

```python
x=df.groupby(['rating']).size().reset_index(name='counts')
fig=px.pie(x,values='counts',names='rating',title='Distribution of Ratings')
fig.show()
```

```python
df['director']=df['director'].fillna('Director Not Specified')
director_list=pd.DataFrame()
director_list=df['director'].str.split(',',expand=True).stack()
director_list=director_list.to_frame()
director_list.columns=['Director']
director_list
```

```python
directors=director_list.groupby(['Director']).size().reset_index(name='Total Count')
directors=directors[directors.Director !='Director Not Specified']
directors=directors.sort_values(by=['Total Count'],ascending=False)
Top5directors=directors.head()
Top5directors=Top5directors.sort_values(by=['Total Count'])
fig=px.bar(Top5directors,x='Total Count',y='Director',title='Top 5 Directors')
fig.show()
```

```python
df['cast']=df['cast'].fillna('Cast Not Specified')
cast_list=pd.DataFrame()
cast_list=df['cast'].str.split(',',expand=True).stack()
cast_list=cast_list.to_frame()
cast_list.columns=['Actor']
actors=cast_list.groupby(['Actor']).size().reset_index(name='Total Count')
actors=actors[actors.Actor !='Cast Not Specified']
actors=actors.sort_values(by=['Total Count'],ascending=False)
Top5Actors=actors.head()
Top5Actors=Top5Actors.sort_values(by=['Total Count'])
fig=px.bar(Top5Actors,x='Total Count',y='Actor',title='Top 5 Actors On Netflix')
fig.show()
```

Analyzing The Content Produced On Netflix Based On Years

```python
df1=df[['type', 'release_year']]
df1=df1.rename(columns={'release_year':'Release Year','type': 'Type'})
df2=df1.groupby(['Release Year','Type']).size().reset_index(name='Total Count')
df2=df2[df2['Release Year']>=2005]
fig=px.line(df2,x='Release Year',y='Total Count',color='Type',title='Content Produced On Netflix Based On Years')
fig.show()
```

Sentiment Analysis Of Netflix Content

```python
df3=df[['release_year','description']]
df3=df3.rename(columns={'release_year':'Release Year','description':'Description'})
for index, row in df3.iterrows():
  d=row['Description']
  testimonial=TextBlob(d)
  p=testimonial.sentiment.polarity
  if p==0:
    sent='Neutral'
  elif p>0:
    sent='Positive'
  else:
    sent='Negative'
  df3.loc[[index,2],'Sentiment']=sent
df3=df3.groupby(['Release Year','Sentiment']).size().reset_index(name='Total Count')
df3=df3[df3['Release Year']>2010]
fig=px.bar(df3,x='Release Year',y='Total Count',color='Sentiment',title='Sentiment Analysis Of Netflix Content')
fig.show()
```
